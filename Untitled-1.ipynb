```python
import cv2 as cv
import numpy as np
import matplotlib.pyplot as plt
from scipy.ndimage import gaussian_laplace

img = cv.imread("a2images/the_berry_farms_sunflower_field.jpeg", cv.IMREAD_REDUCED_COLOR_4)
gray = cv.cvtColor(img, cv.COLOR_BGR2GRAY).astype(np.float32) / 255.0

# Define scale-space range for sigma
sigma_min, sigma_max, num_scales = 2.0, 40.0, 25
k = (sigma_max / sigma_min) ** (1.0 / (num_scales - 1))
sigmas = [sigma_min * (k ** i) for i in range(num_scales)]

# Compute normalized scale-space representation
scale_space = np.zeros((gray.shape[0], gray.shape[1], num_scales), dtype=np.float32)
for i, s in enumerate(sigmas):
    # Scale-normalized LoG: -s^2 * LoG finds bright/dark centered blobs
    scale_space[:, :, i] = np.abs((s ** 2) * gaussian_laplace(gray, sigma=s))

# 3D Non-Maximum Suppression
from scipy.ndimage import maximum_filter
neighborhood = np.ones((3, 3, 3), dtype=bool)
local_max = maximum_filter(scale_space, footprint=neighborhood) == scale_space

threshold = 0.08 * scale_space.max()
peaks = np.argwhere(local_max & (scale_space > threshold))

blobs = []
for y, x, s_idx in peaks:
    s = sigmas[s_idx]
    r = np.sqrt(2) * s
    blobs.append((x, y, r, s, scale_space[y, x, s_idx]))

# Sort by radius descending
blobs.sort(key=lambda b: b[2], reverse=True)
```
