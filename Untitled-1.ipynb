```python
import os
import cv2 as cv
import cv2
import numpy as np
import matplotlib.pyplot as plt
from scipy.optimize import minimize
from scipy import linalg
from scipy.ndimage import gaussian_laplace

# Configure plot styling
plt.rcParams['figure.figsize'] = (10, 8)
plt.rcParams['image.cmap'] = 'gray'# 1. Load sunflower field image
img_path = 'a2images/the_berry_farms_sunflower_field.jpeg'

# Graceful fallback check if image path is local
if os.path.exists(img_path):
    im_bgr = cv.imread(img_path, cv.IMREAD_REDUCED_COLOR_4)
    im_rgb = cv.cvtColor(im_bgr, cv.COLOR_BGR2RGB)
    gray = cv.cvtColor(im_bgr, cv.COLOR_BGR2GRAY).astype(np.float64) / 255.0
else:
    print(f"Warning: {img_path} not found. Generating a synthetic placeholder test image.")
    gray = np.zeros((400, 600), dtype=np.float64)
    cv.circle(gray, (150, 150), 35, 1.0, -1)
    cv.circle(gray, (350, 200), 50, 1.0, -1)
    cv.circle(gray, (480, 120), 20, 1.0, -1)
    im_rgb = np.stack([gray, gray, gray], axis=-1)

# 2. Scale-space parameters
# Sunflowers in the field range from small (distant) to large (foreground)
sigma_min = 0.5
sigma_max = 10.0
num_scales = 20
k = (sigma_max / sigma_min) ** (1.0 / (num_scales - 1))
sigma_values = [sigma_min * (k ** i) for i in range(num_scales)]

# 3. Build Scale-Space LoG Volume
H, W = gray.shape
log_volume = np.zeros((H, W, num_scales), dtype=np.float64)

for i, s in enumerate(sigma_values):
    # Scale-normalized Laplacian of Gaussian: -s^2 * Laplacian(G_s * I)
    # The negative sign captures bright circular blooms on darker green foliage
    response = - (s ** 2) * gaussian_laplace(gray, sigma=s)
    log_volume[:, :, i] = response

# 4. 3D Non-Maximum Suppression
threshold = 0.12 * log_volume.max()
blobs = []

for s_idx in range(1, num_scales - 1):
    slice_curr = log_volume[:, :, s_idx]
    slice_prev = log_volume[:, :, s_idx - 1]
    slice_next = log_volume[:, :, s_idx + 1]
    
    # 2D local max within the current scale
    from scipy.ndimage import maximum_filter
    local_max = maximum_filter(slice_curr, size=3) == slice_curr
    
    # Check scale neighbors
    is_scale_max = (slice_curr > slice_prev) & (slice_curr > slice_next)
    candidates = local_max & is_scale_max & (slice_curr > threshold)
    
    y_coords, x_coords = np.nonzero(candidates)
    s = sigma_values[s_idx]
    r = np.sqrt(2) * s
    
    for y, x in zip(y_coords, x_coords):
        blobs.append((y, x, r, s, slice_curr[y, x]))

# 5. Overlap elimination / 2D spatial NMS
blobs = sorted(blobs, key=lambda b: b[4], reverse=True)
filtered_blobs = []

def overlap(b1, b2):
    y1, x1, r1, _, _ = b1
    y2, x2, r2, _, _ = b2
    d = np.hypot(x1 - x2, y1 - y2)
    return d < (r1 + r2) * 0.5

for b in blobs:
    if not any(overlap(b, fb) for fb in filtered_blobs):
        filtered_blobs.append(b)

# 6. Report largest circles and visualize
sorted_by_size = sorted(filtered_blobs, key=lambda b: b[2], reverse=True)
largest_blobs = sorted_by_size[:10]

print("="*60)
print(f"Scale range (sigma) used: [{sigma_min:.2f}, {sigma_max:.2f}] across {num_scales} scales")
print("Top 10 Largest Detected Circles (Center X, Center Y, Radius r, Sigma):")
for idx, (cy, cx, r, s, score) in enumerate(largest_blobs, 1):
    print(f"{idx:2d}. Center: ({cx:4.1f}, {cy:4.1f}) | Radius: {r:5.2f} px | Sigma: {s:5.2f}")
print("="*60)

fig, ax = plt.subplots(figsize=(12, 10))
ax.imshow(im_rgb)
for (cy, cx, r, s, _) in filtered_blobs:
    c = plt.Circle((cx, cy), r, color='lime', fill=False, linewidth=1.5)
    ax.add_patch(c)

# Highlight top 5 largest circles in red
for (cy, cx, r, s, _) in largest_blobs[:5]:
    c = plt.Circle((cx, cy), r, color='red', fill=False, linewidth=2.5)
    ax.add_patch(c)

ax.set_title("Question 1: Detected Sunflower Blobs (Red = Largest Blooms)", fontsize=14)
ax.axis('off')
plt.tight_layout()
plt.show()
```

# Question 2: Robust Model Fitting using RANSAC (Line and Circle)

```python

import numpy as np
from scipy.optimize import minimize
from scipy import linalg
import matplotlib.pyplot as plt

# np.random.seed(0)

N = 100
half_n = N // 2
r = 10
x0_gt, y0_gt = 2, 3  # Center
s = r / 16
t = np.random.uniform(0, 2 * np.pi, half_n)
n = s * np.random.randn(half_n)
x, y = x0_gt + (r + n) * np.cos(t), y0_gt + (r + n) * np.sin(t)
X_circ = np.hstack((x.reshape(half_n, 1), y.reshape(half_n, 1)))

s = 1.0
m, b = -1, 2
x = np.linspace(-12, 12, half_n)
y = m * x + b + s * np.random.randn(half_n)
X_line = np.hstack((x.reshape(half_n, 1), y.reshape(half_n, 1)))

X = np.vstack((X_circ, X_line))

fig, ax = plt.subplots(1, 1, figsize=(8, 8))
ax.scatter(X_line[:, 0], X_line[:, 1], label='Line')
ax.scatter(X_circ[:, 0], X_circ[:, 1], label='Circle')
circle_gt = plt.Circle((x0_gt, y0_gt), r, color='g', fill=False, label='Ground_truth_circle')
ax.add_patch(circle_gt)
ax.plot(x0_gt, y0_gt, '+', color='g')
x_min, x_max = ax.get_xlim()
x_ = np.array([x_min, x_max])
y_ = m * x_ + b
ax.plot(x_, y_, color='m', label='Ground_truth_line')
ax.legend()
plt.show()
```

```python

# Function to fit a line through 2 points and return unit normal [a, b] and d
def get_line_params(p1, p2):
    diff = p2 - p1
    dist = np.linalg.norm(diff)
    if dist < 1e-7:
        return None
    # Unit normal vector perpendicular to direction vector (diff[0], diff[1])
    a = -diff[1] / dist
    b = diff[0] / dist
    d = a * p1[0] + b * p1[1]
    return a, b, d

# Total Least Squares line refitting on inliers
def tls_line_refit(pts):
    mean = np.mean(pts, axis=0)
    pts_centered = pts - mean
    _, _, vt = np.linalg.svd(pts_centered)
    normal = vt[-1]  # Eigenvector corresponding to smallest singular value
    a, b = normal / np.linalg.norm(normal)
    d = a * mean[0] + b * mean[1]
    return a, b, d

line_thresh = 1.5  # Normal distance threshold (s = 1.0)
n_iters_line = 1000
best_line_inliers = []
best_line_sample_idx = None
best_sample_line = None

for _ in range(n_iters_line):
    idx = np.random.choice(len(X), 2, replace=False)
    params = get_line_params(X[idx[0]], X[idx[1]])
    if params is None:
        continue
    a_cand, b_cand, d_cand = params
    
    # Distance to line: |a*x + b*y - d| / sqrt(a^2 + b^2) = |a*x + b*y - d|
    errors = np.abs(a_cand * X[:, 0] + b_cand * X[:, 1] - d_cand)
    inliers = np.where(errors < line_thresh)[0]
    
    if len(inliers) > len(best_line_inliers):
        best_line_inliers = inliers
        best_line_sample_idx = idx
        best_sample_line = (a_cand, b_cand, d_cand)

# Refit line on all consensus inliers using Total Least Squares
a_fit, b_fit, d_fit = tls_line_refit(X[best_line_inliers])

print(f"Line RANSAC: Found {len(best_line_inliers)} inliers out of {len(X)} points.")


# Question 2 (b): Subtract Consensus and Estimate Circle using RANSAC


# Function to fit an exact circle through 3 non-collinear points
def fit_circle_3pts(p1, p2, p3):
    d = 2 * (p1[0]*(p2[1] - p3[1]) + p2[0]*(p3[1] - p1[1]) + p3[0]*(p1[1] - p2[1]))
    if abs(d) < 1e-7:
        return None
    p1_sq = p1[0]**2 + p1[1]**2
    p2_sq = p2[0]**2 + p2[1]**2
    p3_sq = p3[0]**2 + p3[1]**2
    
    cx = (p1_sq*(p2[1] - p3[1]) + p2_sq*(p3[1] - p1[1]) + p3_sq*(p1[1] - p2[1])) / d
    cy = (p1_sq*(p3[0] - p2[0]) + p2_sq*(p1[0] - p3[0]) + p3_sq*(p2[0] - p1[0])) / d
    rad = np.hypot(p1[0] - cx, p1[1] - cy)
    return cx, cy, rad

# Subtract the consensus of the best line
remnant_idx = np.setdiff1d(np.arange(len(X)), best_line_inliers)
X_remnant = X[remnant_idx]

circle_thresh = 1.2  # Radial error threshold
n_iters_circle = 2000
best_circle_inliers_rem = []
best_circle_sample_idx_rem = None
best_sample_circle = None

for _ in range(n_iters_circle):
    idx = np.random.choice(len(X_remnant), 3, replace=False)
    circ = fit_circle_3pts(X_remnant[idx[0]], X_remnant[idx[1]], X_remnant[idx[2]])
    if circ is None:
        continue
    cx_cand, cy_cand, rad_cand = circ
    
    # Radial error: |sqrt((x - cx)^2 + (y - cy)^2) - rad|
    errors = np.abs(np.hypot(X_remnant[:, 0] - cx_cand, X_remnant[:, 1] - cy_cand) - rad_cand)
    inliers = np.where(errors < circle_thresh)[0]
    
    if len(inliers) > len(best_circle_inliers_rem):
        best_circle_inliers_rem = inliers
        best_circle_sample_idx_rem = idx
        best_sample_circle = (cx_cand, cy_cand, rad_cand)

# Map remnant inlier indices back to original X indices
best_circle_inliers = remnant_idx[best_circle_inliers_rem]
best_circle_sample_pts = X_remnant[best_circle_sample_idx_rem]

# Refine circle fitting on all circle inliers using scipy.optimize.minimize
def circle_loss(params, pts):
    cx, cy, rad = params
    return np.sum((np.hypot(pts[:, 0] - cx, pts[:, 1] - cy) - rad) ** 2)

res = minimize(circle_loss, best_sample_circle, args=(X[best_circle_inliers],), method='Nelder-Mead')
cx_fit, cy_fit, r_fit = res.x

print(f"Circle RANSAC: Found {len(best_circle_inliers)} inliers out of {len(X_remnant)} remnant points.")
print(f"Estimated Circle Center: ({cx_fit:.2f}, {cy_fit:.2f}), Radius: {r_fit:.2f}")


# Question 2 (c): Visualization (Matching Figure 1)

fig, ax = plt.subplots(1, 1, figsize=(9, 9))

# 1. All points
ax.scatter(X[:, 0], X[:, 1], color='#1f77b4', s=28, label='All points')

# 2. Line inliers
ax.scatter(X[best_line_inliers, 0], X[best_line_inliers, 1], color='#bcbd22', s=28, label='Line inliers')

# 3. Best sample for line (2 points)
line_sample_pts = X[best_line_sample_idx]
ax.scatter(line_sample_pts[:, 0], line_sample_pts[:, 1], color='#e377c2', marker='x', s=55, label='Best sample for line')

# 4. Circle inliers
ax.scatter(X[best_circle_inliers, 0], X[best_circle_inliers, 1], color='#17becf', s=28, label='Circle inliers')

# 5. Best sample for circle (3 points)
ax.scatter(best_circle_sample_pts[:, 0], best_circle_sample_pts[:, 1], color='red', s=40, label='Best sample for circle')

# 6. Ground truth circle
circle_gt_plot = plt.Circle((x0_gt, y0_gt), r, color='seagreen', fill=False, linewidth=1.2, label='Ground truth circle')
ax.add_patch(circle_gt_plot)
ax.plot(x0_gt, y0_gt, '+', color='seagreen', markersize=6)

# 7. RANSAC circle (Refitted)
circle_ransac_plot = plt.Circle((cx_fit, cy_fit), r_fit, color='#d62728', fill=False, linewidth=1.2, label='RANSAC circle')
ax.add_patch(circle_ransac_plot)

# 8. Ground truth line (using m and b directly from your snippet)
x_span = np.array([-14, 15])
y_span_gt = m * x_span + b
ax.plot(x_span, y_span_gt, color='#2ca02c', linewidth=1.2, label='Ground truth line')

# 9. RANSAC line (Refitted: a*x + b*y - d = 0 => y = (d - a*x) / b)
y_span_ransac = (d_fit - a_fit * x_span) / b_fit
ax.plot(x_span, y_span_ransac, color='#9467bd', linewidth=1.2, label='RANSAC line')

ax.set_xlim([-15, 15])
ax.set_ylim([-13, 16])
ax.set_aspect('equal')
ax.grid(True, linestyle='--', alpha=0.5)
ax.legend(loc='upper right', framealpha=0.9, fontsize=9)
plt.title("RANSAC Line and Circle Fitting", fontsize=13)
plt.show()
```

# Question 3: Planar Homography & Image Superimposition

```python
import cv2 as cv
import numpy as np
import matplotlib.pyplot as plt

def superimpose_homography(target_img, src_img, dst_quad, alpha=0.85):
    """
    Computes homography and maps src_img onto a 4-point quadrilateral on target_img.
    
    Parameters:
        target_img : Base image in RGB (H, W, 3)
        src_img    : Image to superimpose in RGB (H, W, 3)
        dst_quad   : np.float32 of shape (4, 2) in order [TL, TR, BR, BL]
        alpha      : Blend transparency factor (0.0 to 1.0)
    """
    h_src, w_src = src_img.shape[:2]
    h_tgt, w_tgt = target_img.shape[:2]
    
    # 1. Source image corner coordinates: [Top-Left, Top-Right, Bottom-Right, Bottom-Left]
    src_quad = np.float32([
        [0, 0],
        [w_src - 1, 0],
        [w_src - 1, h_src - 1],
        [0, h_src - 1]
    ])
    
    # 2. Compute 3x3 Projective Homography Matrix
    H = cv.getPerspectiveTransform(src_quad, dst_quad)
    
    # 3. Warp source image and binary mask into the target image space
    warped_src = cv.warpPerspective(src_img, H, (w_tgt, h_tgt))
    mask = np.ones((h_src, w_src), dtype=np.float32)
    warped_mask = cv.warpPerspective(mask, H, (w_tgt, h_tgt))
    mask_3c = np.repeat(warped_mask[:, :, np.newaxis], 3, axis=2)
    
    # 4. Blend warped source over the selected planar target surface
    blended = target_img.astype(np.float32).copy()
    region = mask_3c > 0.5
    blended[region] = alpha * warped_src[region] + (1.0 - alpha) * target_img[region]
    
    return np.clip(blended, 0, 255).astype(np.uint8), H

import cv2 as cv
import numpy as np

def get_points_cv(image_path):
    img = cv.imread(image_path)
    display_img = img.copy()
    pts = []

    def mouse_callback(event, x, y, flags, param):
        if event == cv.EVENT_LBUTTONDOWN:
            pts.append([x, y])
            # Draw point on the window
            cv.circle(display_img, (x, y), 5, (0, 0, 255), -1)
            cv.putText(display_img, f"{len(pts)}", (x + 8, y - 8),
                       cv.FONT_HERSHEY_SIMPLEX, 0.6, (0, 255, 0), 2)
            cv.imshow("Click 4 Points (TL -> TR -> BR -> BL), then press any key", display_img)
            print(f"Point {len(pts)}: [{x}, {y}]")

    win_name = "Click 4 Points (TL -> TR -> BR -> BL), then press any key"
    cv.namedWindow(win_name)
    cv.setMouseCallback(win_name, mouse_callback)
    cv.imshow(win_name, display_img)

    # Wait until 4 points are clicked and a key is pressed
    while True:
        key = cv.waitKey(20) & 0xFF
        if len(pts) == 4 or key == 27 or key == ord('q'):
            break

    cv.waitKey(500)
    cv.destroyAllWindows()
    for _ in range(5): 
        cv.waitKey(1)
        
    return np.float32(pts)


```

# Question 4: PCB Image Alignment & Defect/Difference Detection

```python
def align_and_compare_pcb(template_path, target_path, ratio_thresh=0.75):
    """
    Aligns target PCB image to template PCB image using SIFT + RANSAC Homography
    and calculates difference image.
    """
    if not (os.path.exists(template_path) and os.path.exists(target_path)):
        print(f"Skipping {target_path} (File not found in local workspace).")
        return None
        
    img_ref = cv.imread(template_path)
    img_tgt = cv.imread(target_path)
    
    gray_ref = cv.cvtColor(img_ref, cv.COLOR_BGR2GRAY)
    gray_tgt = cv.cvtColor(img_tgt, cv.COLOR_BGR2GRAY)
    
    # (a) Detect SIFT Features
    sift = cv.SIFT_create()
    kp_ref, des_ref = sift.detectAndCompute(gray_ref, None)
    kp_tgt, des_tgt = sift.detectAndCompute(gray_tgt, None)
    
    # (b) Match Features with BFMatcher and Ratio Test
    bf = cv.BFMatcher(cv.NORM_L2)
    raw_matches = bf.knnMatch(des_tgt, des_ref, k=2)
    
    good_matches = []
    for m, n in raw_matches:
        if m.distance < ratio_thresh * n.distance:
            good_matches.append(m)
            
    # Extract locations of matched keypoints
    src_pts = np.float32([kp_tgt[m.queryIdx].pt for m in good_matches]).reshape(-1, 1, 2)
    dst_pts = np.float32([kp_ref[m.trainIdx].pt for m in good_matches]).reshape(-1, 1, 2)
    
    # Find Homography using RANSAC
    H, mask = cv.findHomography(src_pts, dst_pts, cv.RANSAC, 5.0)
    
    # Draw top matches
    matches_vis = cv.drawMatches(img_tgt, kp_tgt, img_ref, kp_ref, good_matches[:60],
                                 None, flags=cv.DrawMatchesFlags_NOT_DRAW_SINGLE_POINTS)
    
    # (c) Warp target PCB to align with template
    h_ref, w_ref = gray_ref.shape
    aligned_tgt = cv.warpPerspective(img_tgt, H, (w_ref, h_ref))
    
    # Compute Absolute Difference
    diff = cv.absdiff(img_ref, aligned_tgt)
    diff_gray = cv.cvtColor(diff, cv.COLOR_BGR2GRAY)
    
    # Colorized heat map for visual clarity
    diff_heatmap = cv.applyColorMap(diff_gray, cv.COLORMAP_JET)
    
    return {
        'matches_vis': cv.cvtColor(matches_vis, cv.COLOR_BGR2RGB),
        'ref': cv.cvtColor(img_ref, cv.COLOR_BGR2RGB),
        'aligned': cv.cvtColor(aligned_tgt, cv.COLOR_BGR2RGB),
        'diff': cv.cvtColor(diff, cv.COLOR_BGR2RGB),
        'diff_heatmap': cv.cvtColor(diff_heatmap, cv.COLOR_BGR2RGB)
    }

# Template and candidate PCB images list
template_file = 'a2images/pcbs/AECC1.jpg'
pcb_files = ['a2images/pcbs/AECC2.jpg', 'a2images/pcbs/AECC3.jpg', 'a2images/pcbs/AECC4.jpg']

# Execute alignment across available files
for pcb_file in pcb_files:
    res = align_and_compare_pcb(template_file, pcb_file)
    if res is not None:
        fig, axes = plt.subplots(1, 4, figsize=(20, 5))
        axes[0].imshow(res['ref'])
        axes[0].set_title(f"Template ({template_file})")
        axes[0].axis('off')
        
        axes[1].imshow(res['aligned'])
        axes[1].set_title(f"Aligned Target ({pcb_file})")
        axes[1].axis('off')
        
        axes[2].imshow(res['diff'])
        axes[2].set_title("Difference Image")
        axes[2].axis('off')
        
        axes[3].imshow(res['diff_heatmap'])
        axes[3].set_title("Difference Heatmap")
        axes[3].axis('off')
        
        plt.tight_layout()
        plt.show()
```

e) 2D homography assumes that we use planes. Since some components are relatively tall, the small perspective change affects the image so that it cant be matched using 2d homography (misalligning the top part of these). Sometimes since the more metallic componants are very reflective, then even under identical light slight misallignment might have different reflections which isnt a property changed by applying this transform.
